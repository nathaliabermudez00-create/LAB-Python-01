```python
def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    notas.clear()

mis_notas = [3.5, 4.0]

print("Lista inicial:", mis_notas)
print("ID inicial:", id(mis_notas))

agregar_nota(mis_notas, 4.8)

print("Después de agregar:", mis_notas)
print("ID después de agregar:", id(mis_notas))

reiniciar(mis_notas)

print("Después de reiniciar:", mis_notas)
print("ID después de reiniciar:", id(mis_notas))
```

```python
cuenta = {
    "titular": "Ana",
    "saldo": 200000,
    "movimientos": []
}

def consultar(cuenta):
    print("Titular:", cuenta["titular"])
    print("Saldo actual:", cuenta["saldo"])

def depositar(cuenta, monto):
    if monto > 0:
        cuenta["saldo"] += monto

        movimiento = {
            "tipo": "Depósito",
            "monto": monto,
            "saldo": cuenta["saldo"]
        }

        cuenta["movimientos"].append(movimiento)
        print("Depósito realizado correctamente.")
    else:
        print("El monto debe ser mayor a cero.")

def retirar(cuenta, monto):
    if monto <= 0:
        print("El monto debe ser mayor a cero.")
    elif monto > cuenta["saldo"]:
        print("Saldo insuficiente.")
    else:
        cuenta["saldo"] -= monto

        movimiento = {
            "tipo": "Retiro",
            "monto": monto,
            "saldo": cuenta["saldo"]
        }

        cuenta["movimientos"].append(movimiento)
        print("Retiro realizado correctamente.")

opcion = ""

while opcion != "salir":
    print("\n--- CAJERO AUTOMÁTICO ---")
    print("1. Consultar saldo")
    print("2. Depositar dinero")
    print("3. Retirar dinero")
    print("4. Salir")

    opcion = input("Seleccione una opción: ").lower()

    if opcion == "1":
        consultar(cuenta)

    elif opcion == "2":
        monto = float(input("Ingrese el monto a depositar: "))
        depositar(cuenta, monto)

    elif opcion == "3":
        monto = float(input("Ingrese el monto a retirar: "))
        retirar(cuenta, monto)

    elif opcion == "4" or opcion == "salir":
        opcion = "salir"

    else:
        print("Opción inválida.")

print("\n--- HISTORIAL DE MOVIMIENTOS ---")

for movimiento in cuenta["movimientos"]:
    print(
        movimiento["tipo"],
        "| Monto:", movimiento["monto"],
        "| Saldo:", movimiento["saldo"]
    )

print("Saldo final:", cuenta["saldo"])
```
