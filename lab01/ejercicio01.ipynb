```python
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el valor del depósito: "))
retiro = float(input("Ingrese el valor del retiro: "))

saldo = saldo + deposito
saldo = saldo - retiro

interes = 0.02
saldo_final = saldo + (saldo * interes)

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

temperaturas.append(26.0)

temperaturas.pop(0)

temperaturas.insert(1, 23.7)

temperaturas.sort()
print("Temperaturas ordenadas:", temperaturas)

diferencia = max(temperaturas) - min(temperaturas)
print("Diferencia:", diferencia)

suma = 0
for temperatura in temperaturas:
    suma += temperatura

promedio = suma / len(temperaturas)
print("Promedio:", promedio)

if 23.7 in temperaturas:
    print("Índice de 23.7:", temperaturas.index(23.7))
else:
    print("La temperatura 23.7 no está en la lista")

print("Segunda a cuarta posición:", temperaturas[1:4])

print("Tres primeras temperaturas:", temperaturas[:3])
```

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10}
]

print("Primer producto:", inventario[0]["nombre"])

nuevo_producto = {"nombre": "Arroz", "precio": 4000, "stock": 8}
inventario.append(nuevo_producto)

for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2

total = 0
for producto in inventario:
    total += producto["precio"] * producto["stock"]

print("Valor total:", total)

print("Productos con stock menor a 10:")
for producto in inventario:
    if producto["stock"] < 10:
        print(producto["nombre"])

for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)

nombre_buscar = input("Ingrese el nombre del producto: ")
encontrado = False

for producto in inventario:
    if producto["nombre"].lower() == nombre_buscar.lower():
        print("Producto encontrado:", producto)
        encontrado = True

if encontrado == False:
    print("Producto no encontrado")

i = 0
while i < len(inventario):
    if inventario[i]["stock"] == 0:
        inventario.pop(i)
    else:
        i += 1

for i in range(len(inventario)):
    for j in range(i + 1, len(inventario)):
        if inventario[i]["precio"] > inventario[j]["precio"]:
            temporal = inventario[i]
            inventario[i] = inventario[j]
            inventario[j] = temporal

print("Inventario final:", inventario)
```

```python
# Tupla de coordenadas
punto = (3, 7)
print("Punto:", punto)

# Crear una tupla para el rectángulo
ancho = 5
alto = 3
rectangulo = (ancho, alto)

# Calcular área y perímetro
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])

print("Área:", area)
print("Perímetro:", perimetro)

# Intentar cambiar el ancho
try:
    rectangulo[0] = 10
except TypeError:
    print("Error: las tuplas no permiten modificar sus elementos")

# Lista de posiciones
posiciones = [(0, 0), (1, 2), (3, 4)]

# Calcular distancia total recorrida
distancia_total = 0

for i in range(len(posiciones) - 1):
    x1 = posiciones[i][0]
    y1 = posiciones[i][1]
    x2 = posiciones[i + 1][0]
    y2 = posiciones[i + 1][1]

    distancia = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
    distancia_total += distancia

print("Distancia total:", distancia_total)
```

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True}
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

prestados = 0
disponibles = 0

for libro in biblioteca["libros"]:
    if libro["prestado"] == True:
        prestados += 1
    else:
        disponibles += 1

print("Libros prestados:", prestados)
print("Libros disponibles:", disponibles)

nuevo_libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False
}

biblioteca["libros"].append(nuevo_libro)

for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True

print("Biblioteca actualizada:")
print(biblioteca)
```
