```python
import random

ganadas = 0
perdidas = 0
jugar = "s"

while jugar == "s":

    numero_secreto = random.randint(1, 100)
    historial = []
    encontrado = False

    print("\nAdivina el número entre 1 y 100.")
    print("Tienes 7 intentos.")

    for intento in range(7):
        numero = int(input("Escribe un número: "))
        historial.append(numero)

        if numero == numero_secreto:
            print("¡Correcto! Adivinaste el número.")
            encontrado = True
            ganadas += 1
            break

        elif numero < numero_secreto:
            print("El número secreto es mayor.")

        else:
            print("El número secreto es menor.")

    if encontrado == False:
        print("Se acabaron los intentos.")
        print("El número secreto era:", numero_secreto)
        perdidas += 1

    print("Historial de intentos:", historial)
    print("Partidas ganadas:", ganadas)
    print("Partidas perdidas:", perdidas)

    jugar = input("¿Quieres jugar otra vez? (s/n): ").lower()

print("\nFin del juego.")
print("Total de partidas ganadas:", ganadas)
print("Total de partidas perdidas:", perdidas)
```

```python
def contar_vecinos(tablero, fila, columna):
    cantidad = 0

    for cambio_fila in range(-1, 2):
        for cambio_columna in range(-1, 2):

            if cambio_fila == 0 and cambio_columna == 0:
                continue

            nueva_fila = fila + cambio_fila
            nueva_columna = columna + cambio_columna

            if nueva_fila >= 0 and nueva_fila < 10:
                if nueva_columna >= 0 and nueva_columna < 10:
                    if tablero[nueva_fila][nueva_columna] == 1:
                        cantidad += 1

    return cantidad


def siguiente_generacion(tablero):
    nuevo_tablero = []

    for fila in range(10):
        nueva_fila = []

        for columna in range(10):
            nueva_fila.append(0)

        nuevo_tablero.append(nueva_fila)

    for fila in range(10):
        for columna in range(10):

            vecinos = contar_vecinos(tablero, fila, columna)

            if tablero[fila][columna] == 1:
                if vecinos == 2 or vecinos == 3:
                    nuevo_tablero[fila][columna] = 1
                else:
                    nuevo_tablero[fila][columna] = 0

            else:
                if vecinos == 3:
                    nuevo_tablero[fila][columna] = 1

    return nuevo_tablero


def mostrar_tablero(tablero):
    for fila in range(10):
        for columna in range(10):

            if tablero[fila][columna] == 1:
                print("■", end=" ")
            else:
                print("·", end=" ")

        print()


# Crear tablero de 10 x 10
tablero = []

for fila in range(10):
    nueva_fila = []

    for columna in range(10):
        nueva_fila.append(0)

    tablero.append(nueva_fila)


# Crear figura inicial: tres células vivas horizontales
tablero[4][3] = 1
tablero[4][4] = 1
tablero[4][5] = 1


# Mostrar cinco generaciones
for generacion in range(5):
    print("\nGeneración:", generacion)
    mostrar_tablero(tablero)

    tablero = siguiente_generacion(tablero)
```
