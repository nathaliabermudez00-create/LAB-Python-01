```python
productos = []

while True:
    nombre = input("Nombre del producto (o escriba fin): ")

    if nombre.lower() == "fin":
        break

    precio_valido = False

    while precio_valido == False:
        try:
            precio = float(input("Precio del producto: "))

            if precio < 0:
                print("El precio no puede ser negativo.")
            else:
                precio_valido = True

        except ValueError:
            print("Error: debe ingresar un número válido.")

    cantidad_valida = False

    while cantidad_valida == False:
        try:
            cantidad = int(input("Cantidad del producto: "))

            if cantidad < 0:
                print("La cantidad no puede ser negativa.")
            else:
                cantidad_valida = True

        except ValueError:
            print("Error: debe ingresar un número entero.")

    producto = {
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    }

    productos.append(producto)

print("\nInventario registrado:")

for producto in productos:
    print(producto)


# Ejemplo de uso de else y finally
try:
    print("\nCantidad de productos:", len(productos))

except Exception:
    print("Ocurrió un error.")

else:
    print("El inventario se procesó correctamente.")

finally:
    print("Proceso de inventario terminado.")
```

```python
import json


def guardar_csv(productos):
    archivo = open("inventario.csv", "w")

    for producto in productos:
        linea = (
            producto["nombre"] + ","
            + str(producto["precio"]) + ","
            + str(producto["cantidad"]) + "\n"
        )

        archivo.write(linea)

    archivo.close()
    print("Inventario guardado en CSV.")


def leer_csv():
    productos_leidos = []

    try:
        archivo = open("inventario.csv", "r")

        for linea in archivo:
            datos = linea.strip().split(",")

            producto = {
                "nombre": datos[0],
                "precio": float(datos[1]),
                "cantidad": int(datos[2])
            }

            productos_leidos.append(producto)

        archivo.close()

    except FileNotFoundError:
        print("El archivo CSV no existe. Se inicia un inventario vacío.")

    return productos_leidos


def guardar_json(productos):
    archivo = open("inventario.json", "w")

    json.dump(productos, archivo, indent=4)

    archivo.close()
    print("Inventario guardado en JSON.")


def leer_json():
    try:
        archivo = open("inventario.json", "r")

        productos_leidos = json.load(archivo)

        archivo.close()

        return productos_leidos

    except FileNotFoundError:
        print("El archivo JSON no existe. Se inicia un inventario vacío.")
        return []


# Guardar inventario en CSV
guardar_csv(productos)

# Leer inventario desde CSV
inventario_csv = leer_csv()

print("\nInventario recuperado desde CSV:")
for producto in inventario_csv:
    print(producto)


# Guardar inventario en JSON
guardar_json(productos)

# Leer inventario desde JSON
inventario_json = leer_json()

print("\nInventario recuperado desde JSON:")
for producto in inventario_json:
    print(producto)
```
