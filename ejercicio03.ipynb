```python

import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()

nombres = []

for usuario in usuarios:
    correo = usuario["email"]

    if correo.endswith(".biz"):
        nombres.append(usuario["name"])

print("Usuarios con correo terminado en .biz:")

for nombre in nombres:
    print("-", nombre)
```

```python

import requests

nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()

url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()

    tipos = []
    habilidades = []

    for tipo in datos["types"]:
        tipos.append(tipo["type"]["name"])

    for habilidad in datos["abilities"]:
        habilidades.append(habilidad["ability"]["name"])

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades,
        "experiencia": datos["base_experience"]
    }

    print("\nInformación del Pokémon")
    print("Nombre:", pokemon["nombre"].capitalize())
    print("Altura:", pokemon["altura"])
    print("Peso:", pokemon["peso"])

    print("Tipos:")
    for tipo in pokemon["tipos"]:
        print("-", tipo)

    if len(pokemon["tipos"]) > 1:
        print("Pokémon de tipo mixto")
    else:
        print("Pokémon de un solo tipo")

    if pokemon["peso"] > 100:
        print("Pokémon de gran tamaño")
    else:
        print("Pokémon ligero")

    print("Habilidades:")
    for habilidad in pokemon["habilidades"]:
        print("-", habilidad)

    print("Experiencia base:", pokemon["experiencia"])

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
