```python
n = int(input("Ingrese un número: "))

suma = 0

for i in range(1, n + 1):
    suma += i

print("La suma es:", suma)

print("Números pares:")

for i in range(1, n + 1):
    if i % 2 == 0:
        print(i)

temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1, 23.7]

suma_temperaturas = 0

for temperatura in temperaturas:
    suma_temperaturas += temperatura

promedio = suma_temperaturas / len(temperaturas)

print("Promedio:", round(promedio, 2))
print("Temperaturas mayores al promedio:")

for temperatura in temperaturas:
    if temperatura > promedio:
        print(temperatura)
```

```python
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("Estado final:", estado_final)
```

```python
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")

    if nombre.lower() == "fin":
        break

    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    if precio < 0 or cantidad < 0:
        print("Datos inválidos. Producto no agregado.")
        continue

    productos.append({
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    })

total = 0

for producto in productos:
    total += producto["precio"] * producto["cantidad"]

print("Total del inventario:", total)

if len(productos) > 0:
    producto_caro = productos[0]

    for producto in productos:
        if producto["precio"] > producto_caro["precio"]:
            producto_caro = producto

    print("Producto más caro:", producto_caro)
else:
    print("No hay productos en el inventario")

print("Estado final del inventario:")
print(productos)
```
